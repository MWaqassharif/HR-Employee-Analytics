**Check_outlier**

```python
df["MonthlyIncome"].describe()
```

```python
num_cols = df.select_dtypes(include = ["number"])

result = []

for col in num_cols.columns:
    Q1 = np.percentile(df[col], 25)
    Q3 = np.percentile(df[col], 75)
    IQR = Q3 - Q1
    
    upper = Q3 + 1.5*IQR
    lower = Q1 - 1.5*IQR

    outliers = ((df[col]< lower)|(df[col]> upper)).sum()
    
    result.append([col,  outliers])

outlier_df = pd.DataFrame(result, columns =["Column" , "Outliers"] )

outlier_df   
  
    
```

```python
outlier_cols = outlier_df[outlier_df["Outliers"] > 0]["Column"]
outlier_cols
```

**Outliers Observation**

1. MonthlyIncome → 114
2. NumCompaniesWorked → 52
3. PerformanceRating → 226
4. StockOptionLevel → 85
5. TotalWorkingYears → 63
6. TrainingTimesLastYear → 238
7. YearsAtCompany → 104
8. YearsInCurrentRole → 21
9. YearsSinceLastPromotion → 107
10. YearsWithCurrManager → 14

```python
for col in outlier_cols:

    Q1 = np.percentile(df[col], 25)
    Q3 = np.percentile(df[col], 75)

    IQR = Q3 - Q1

    upper = Q3 + 1.5 * IQR
    lower = Q1 - 1.5 * IQR

    investigate = df.loc[(df[col] < lower) | (df[col] > upper),["Age", "JobRole", "JobLevel", col]]

    print("="*70)
    print(f"Column: {col}")
    print(f"Total Outliers: {len(investigate)}")
    display(investigate.head(10))
```

**Observation:**

Outliers were detected in several numerical columns using the IQR method. After investigation, it was found that these values belong to senior employees (such as Managers and Research Directors) or represent valid business information (such as performance ratings and years of experience). Therefore, the detected outliers are realistic and no outlier treatment was applied

## Project Summary

This project analyzed the HR Employee Attrition dataset to understand the main reasons why employees leave the company. Out of **1,470** employees, **237** employees **(16.12%)** left the company, while 1,233 employees **(83.88%)** stayed.

The analysis found that employees who worked overtime, had lower job satisfaction, poor work-life balance, and fewer years at the company were more likely to leave. The Sales department had the highest attrition, and employees aged **26–33** years showed the highest number of resignations. It was also observed that employees with higher education levels earned higher average salaries.

## Business Recommendations

1. Overtime → Reduce overtime because employees working overtime were more likely to leave.
2. Salary → Better pay can improve employee retention.
3. Job Satisfaction → Happy employees are less likely to leave.
4. New Employees → Support them with training and career guidance.
5. Sales Department → Investigate department-specific issues and improve the work environment.

These recommendations can help HR reduce employee attrition, improve employee satisfaction, and increase long-term retention.
